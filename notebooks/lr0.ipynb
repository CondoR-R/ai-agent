## labs/lr0/train.py

```python
import torch
from torch.utils.data import DataLoader, TensorDataset
from ai_agent.labs.lr0 import train, model
from ai_agent.reproducibility import make_generator

features_train, labels_train = features[split['train']], labels[split['train']]
features_t = torch.from_numpy(features_train)
labels_t = torch.from_numpy(labels_train).long()

dataset = TensorDataset(features_t, labels_t)

m = model.MLP(1, 64, 0)
loader = DataLoader(dataset, batch_size=64, shuffle=True, generator=make_generator(0))
train.train(m, {'lr': 0.001, 'weight_decay': 0.0, 'batch_size': 64, 'epochs': 30}, loader)

```
